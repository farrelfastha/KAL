# Perkalian titik (Dot product) dan Vektor satuan

## Perkalian titik (dot product)
Perkalian titik, atau dot product, adalah operasi matematika yang mengalikan dua vektor dan menghasilkan sebuah skalar (bilangan real). Operasi ini mengukur sejauh mana dua vektor "searah" atau "sejajar" satu sama lain.
Jika dua vektor a dan b memiliki komponen sebagai berikut:

$a = (a₁, a₂, ..., aₙ)$

$b = (b₁, b₂, ..., bₙ)$

Maka, perkalian titiknya adalah:

$a · b = a₁b₁ + a₂b₂ + ... + aₙbₙ$

Artinya, kita mengalikan komponen yang bersesuaian dari kedua vektor dan menjumlahkannya.

contoh program untuk menghituung dot product :

```python
def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("Panjang vektor harus sama.")
    return sum(x * y for x, y in zip(a, b))

# Contoh penggunaan
v1 = [1, 3, -5]
v2 = [4, -2, -1]
hasil = dot_product(v1, v2)
print("Dot product:", hasil)  
```

## vektor satuan
Vektor satuan adalah vektor yang memiliki panjang (magnitudo) sebesar 1 satuan dan digunakan untuk menunjukkan arah tertentu dalam ruang. Vektor ini tidak memiliki satuan fisik karena fungsinya hanya untuk menunjukkan arah, bukan besarannya.


Vektor satuan digunakan untuk:

*   Menentukan arah suatu vektor dalam ruang.
*   Membentuk basis dalam ruang vektor, memungkinkan representasi vektor lain sebagai kombinasi linear dari vektor satuan.
*   Mempermudah perhitungan dalam operasi vektor seperti proyeksi, dot product, dan cross product.


Misalkan terdapat vektor A dengan komponen:

$A = 3î + 4ĵ$

Untuk menemukan vektor satuan yang searah dengan A, kita lakukan normalisasi:

Hitung magnitudo A:

$|A| = √(3² + 4²) = √(9 + 16) = √25 = 5$

Bagi setiap komponen dengan magnitudo:

$Â = (3/5)î + (4/5)ĵ$

Sehingga, Â adalah vektor satuan yang searah dengan A.


buat program unutk mengkonversi vektor menjadi vektor satuan (ventor memiliki panjang vektor 1):

```python
import math

def normalize(vector):
    magnitude = math.sqrt(sum(comp**2 for comp in vector))
    if magnitude == 0:
        raise ValueError("Vektor nol tidak dapat dinormalisasi.")
    return [comp / magnitude for comp in vector]

# Contoh penggunaan
v = [3, 4]
unit_v = normalize(v)
print("Vektor satuan:", unit_v)
```
