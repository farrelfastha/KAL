# singular value decomposition SVD
## apa itu SVD
SVD adalah teknik dalam aljabar linier yang digunakan untuk menguraikan matriks menjadi tiga matriks yang lebih sederhana. Ini adalah alat penting untuk reduksi data dan reduksi dimensi, yang sering digunakan dalam berbagai aplikasi seperti pemrosesan gambar, pemodelan topik, dan analisis data. SVD juga dapat digunakan untuk menemukan solusi untuk sistem linear dan masalah kuadrat terkecil, serta menentukan peringkat numerik dari suatu matriks.
## kegunaan SVD
SVD berguna untuk memecah matriks besar menjadi komponen yang lebih sederhana, sehingga bisa dimanfaatkan untuk:

* menyederhanakan data (reduksi dimensi)
* membersihkan noise
* menyusun rekomendasi
* mengkompresi file
* menganalisis pola hubungan antar data
## bagaimana formula SVD
Formula untuk Singular Value Decomposition (SVD) adalah A = UΣVᵀ, di mana A adalah matriks yang akan difaktorkan, U adalah matriks ortogonal kiri, Σ adalah matriks diagonal (dengan nilai singular pada diagonal), dan Vᵀ adalah transpos matriks ortogonal kanan.

Penjelasan:

A (mxn): Matriks asli yang ingin diuraikan, dengan m baris dan n kolom.
U (mxm): Matriks ortogonal kiri. Kolom-kolomnya adalah vektor singular kiri dari A.
Σ (mxn): Matriks diagonal, elemen diagonalnya adalah nilai singular dari A. Nilai-nilai singular ini adalah akar kuadrat dari nilai eigen dari AᵀA (atau AAᵀ).
Vᵀ (nxn): Transpos matriks ortogonal kanan. Baris-barisnya adalah vektor singular kanan dari A.

Kesimpulan: SVD memecah matriks A menjadi tiga komponen: matriks U yang berisi informasi tentang transformasi rotasi (vektor singular kiri), matriks Σ yang berisi informasi tentang skala (nilai singular), dan matriks Vᵀ yang berisi informasi tentang transformasi rotasi lainnya (vektor singular kanan).

Contoh:
Jika A adalah matriks 3x2, maka:  
U akan menjadi matriks 3x3 ortogonal.
Σ akan menjadi matriks 3x2 dengan nilai singular pada diagonal utamanya.
Vᵀ akan menjadi matriks 2x2 ortogonal.


## soal 1


$$
A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$


$$
A = U \Sigma V^T
$$

Langkah 1: Hitung $A^T A$ dan $A A^T$

Karena $A$ adalah simetris, $A^T = A$. Maka:

$$
A^T A = A A =
\begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
\begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
= \begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}
$$

Langkah 2: Cari Eigenvalue dari $A^T A$

Misalkan:

$$
A^T A = M = \begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}
$$

Cari karakteristik polinomial:

$$
\det(M - \lambda I) = 0
$$

$$
\begin{vmatrix}
10 - \lambda & 6 \\
6 & 10 - \lambda
\end{vmatrix} = (10 - \lambda)^2 - 36 = 0
$$

$$
(10 - \lambda)^2 = 36 \Rightarrow 10 - \lambda = \pm 6
\Rightarrow \lambda = 4, 16
$$

Langkah 3: Cari Singular Values

$$
\sigma_1 = \sqrt{16} = 4,\quad \sigma_2 = \sqrt{4} = 2
$$

Langkah 4: Cari Eigenvector dari $A^T A$

Untuk $\lambda = 16$:

$$
(M - 16I) v = 0 \Rightarrow
\begin{bmatrix}
-6 & 6 \\
6 & -6
\end{bmatrix}
\Rightarrow v_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

Untuk $\lambda = 4$:

$$
(M - 4I) v = 0 \Rightarrow
\begin{bmatrix}
6 & 6 \\
6 & 6
\end{bmatrix}
\Rightarrow v_2 = \begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$

Lalu normalisasi:

$$
v_1 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \quad
v_2 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$

Langkah 5: Bangun Matriks $V$ dan $V^T$

$$
V = \left[ v_1\ v_2 \right] =
\frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}
,\quad V^T = V^T
$$

Langkah 6: Hitung $U$ menggunakan $u_i = \frac{1}{\sigma_i} A v_i$

Hitung $u_1$:

$$
u_1 = \frac{1}{4} A v_1 = \frac{1}{4} A \cdot \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
= \frac{1}{4 \sqrt{2}} \begin{bmatrix} 3+1 \\ 1+3 \end{bmatrix}
= \frac{1}{4 \sqrt{2}} \begin{bmatrix} 4 \\ 4 \end{bmatrix}
= \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

Hitung $u_2$:

$$
u_2 = \frac{1}{2} A v_2 = \frac{1}{2} A \cdot \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}
= \frac{1}{2 \sqrt{2}} \begin{bmatrix} 3 - 1 \\ 1 - 3 \end{bmatrix}
= \frac{1}{2 \sqrt{2}} \begin{bmatrix} 2 \\ -2 \end{bmatrix}
= \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ -1 \end{bmatrix}
$$

Langkah 7: Susun Matriks $U$, $\Sigma$, dan hasil akhir

$$
U = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}, \quad
\Sigma = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}, \quad
V^T = \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix}
$$

Jadi SVD dari $A$ adalah:

$$
A = U \Sigma V^T
= \left( \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix} \right)
\begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}
\left( \frac{1}{\sqrt{2}} \begin{bmatrix}
1 & 1 \\
1 & -1
\end{bmatrix} \right)
$$

Tes apakah $U \Sigma V^T = A$:

$$
U \Sigma =
\frac{1}{\sqrt{2}} \begin{bmatrix} 1 & 1 \\ 1 & -1 \end{bmatrix}
\begin{bmatrix} 4 & 0 \\ 0 & 2 \end{bmatrix}
=
\frac{1}{\sqrt{2}} \begin{bmatrix} 4 & 2 \\ 4 & -2 \end{bmatrix}
$$

$$
U \Sigma V^T =
\left( \frac{1}{\sqrt{2}} \begin{bmatrix} 4 & 2 \\ 4 & -2 \end{bmatrix} \right)
\cdot \frac{1}{\sqrt{2}} \begin{bmatrix} 1 & 1 \\ 1 & -1 \end{bmatrix}
=
\frac{1}{2} \begin{bmatrix}
4 + 2 & 4 - 2 \\
4 - 2 & 4 + 2
\end{bmatrix}
=
\begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
= A
$$


## soal 2
$$
A = \begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
$$

Kita ingin mendapatkan dekomposisi:

$$
A = U \Sigma V^T
$$

Hitung $A^T A$

$$
A^T =
\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}
$$

$$
A^T A =
\begin{bmatrix}
3 & 2 & 4 & 1 \\
7 & 5 & 3 & 1
\end{bmatrix}
\cdot
\begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
=
\begin{bmatrix}
30 & 44 \\
44 & 84
\end{bmatrix}
$$

Hitung *eigenvalue* dari $A^T A$

Kita cari nilai $\lambda$ dari:

$$
\det(A^T A - \lambda I) = 0
$$

$$
\left|
\begin{array}{cc}
30 - \lambda & 44 \\
44 & 84 - \lambda
\end{array}
\right|
=
(30 - \lambda)(84 - \lambda) - 44^2 = 0
$$

$$
\Rightarrow \lambda^2 - 114\lambda + (2520 - 1936) = 0
\Rightarrow \lambda^2 - 114\lambda + 584 = 0
$$

Dengan menyelesaikan numerik (karena bentuk tak sederhana):

$$
\lambda_1 \approx 108.62,\quad \lambda_2 \approx 5.38
$$

Hitung *singular value*

$$
\sigma_1 = \sqrt{108.62} \approx 10.42,\quad \sigma_2 = \sqrt{5.38} \approx 2.32
$$

Hitung *vektor eigen (V)* dari $A^T A$

Misalnya:

* Untuk $\lambda_1 = 108.62$, diperoleh $v_1 = \begin{bmatrix} 0.488 \\ 0.873 \end{bmatrix}$
* Untuk $\lambda_2 = 5.38$, diperoleh $v_2 = \begin{bmatrix} -0.873 \\ 0.488 \end{bmatrix}$

Gabung jadi:

$$
V =
\begin{bmatrix}
0.488 & -0.873 \\
0.873 & 0.488
\end{bmatrix},\quad
V^T =
\begin{bmatrix}
0.488 & 0.873 \\
-0.873 & 0.488
\end{bmatrix}
$$

Hitung kolom $U$ dari $u_i = \frac{Av_i}{\sigma_i}$

Contoh:

Untuk $v_1 = \begin{bmatrix} 0.488 \\ 0.873 \end{bmatrix}$:

$$
Av_1 = A \cdot v_1 =
\begin{bmatrix}
3 & 7 \\
2 & 5 \\
4 & 3 \\
1 & 1
\end{bmatrix}
\cdot
\begin{bmatrix}
0.488 \\
0.873
\end{bmatrix}
=
\begin{bmatrix}
7.343 \\
5.982 \\
4.401 \\
1.361
\end{bmatrix}
$$

$$
u_1 = \frac{1}{\sigma_1} Av_1 = \frac{1}{10.42}
\begin{bmatrix}
7.343 \\
5.982 \\
4.401 \\
1.361
\end{bmatrix}
\approx
\begin{bmatrix}
-0.270 \\
-0.150 \\
-0.928 \\
-0.210
\end{bmatrix}
$$

Lakukan hal yang sama untuk $v_2$ → dapatkan $u_2$

Bentuk matriks $\Sigma$

Karena $A$ berukuran $4 \times 2$, maka:

$$
\Sigma =
\begin{bmatrix}
10.42 & 0 \\
0 & 2.32 \\
0 & 0 \\
0 & 0
\end{bmatrix}
\quad \text{(ukuran } 4 \times 2 \text{)}
$$

Rekonstruksi $A = U \Sigma V^T$

Hasil akhir rekontruksi (dengan pembulatan):

$$
A \approx
\begin{bmatrix}
0.16 & -3.31 \\
0.40 & -2.01 \\
-5.34 & -8.09 \\
