# **Perkalian Silang (Cross Product)**

Kali silang atau cross product adalah operasi pada dua vektor dalam ruang tiga dimensi (3D) yang menghasilkan sebuah vektor baru yang tegak lurus (ortogonal) terhadap kedua vektor tersebut.

## contoh perhitungan kali silang

Diketahui dua vektor :
$
\mathbf{A} = (4, -2, 1), \qquad \mathbf{B} = (0, 3, -5)
$






Langkah-langkah:

$% Langkah 1
\textbf{Langkah 1: Tulis Rumus Determinan}
\
\\
\mathbf{A} \times \mathbf{B} =
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
A_x & A_y & A_z \\
B_x & B_y & B_z \\
\end{vmatrix}
\\[10pt]$

\\

$% Langkah 2
\textbf{Langkah 2: Masukkan Nilai Vektor}
\
\\
\mathbf{A} \times \mathbf{B} =
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
4 & -2 & 1 \\
0 & 3  & -5 \\
\end{vmatrix}$

\\

$% Langkah 3
\textbf{Langkah 3: Hitung Masing-Masing Komponen}$

Komponen  $\mathbf{i}:
(-2 \times -5) - (1 \times 3) = 10 - 3 = 7
$

Komponen $\mathbf{j}:
\left( (4 \times -5) - (1 \times 0) \right)
= - ( -20 - 0 ) = 20
$

Komponen $\mathbf{k}:
\
(4 \times 3) - ( -2 \times 0 ) = 12 - 0 = 12
$

\\

$% Langkah 4
\textbf{Langkah 4: Tulis Hasil Akhir}$

$
\mathbf{A} \times \mathbf{B} = (7, 20, 12)
$


# Tugas Soal
## soal
1. tentukan luas jajar genjang yang ditentukan oleh vektor $$\vec{u} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}, \quad \vec{v} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}$$
2. tentukan luas jajar genjang yang ditentukan oleh vektor u
$$
\vec{u} = \begin{bmatrix} 2 \\ 0 \end{bmatrix}, \quad 
\vec{v} = \begin{bmatrix} 0 \\ 3 \end{bmatrix}
$$

3. tentukan luas segitiga dengan titik sudut (0,0,0), (1,3,-1), dan (2,1,1)
4. tentukan luas segitiga dengan titik sudut (5,2,-1), (3,6,2), dan (1,0,4)

## jawaban
### 1. 
$$
L = \left| \det \begin{pmatrix} 1 & 2 \\ 2 & 1 \end{pmatrix} \right| 
= |(1)(1) - (2)(2)| = |1 - 4| = \boxed{3}
$$

---

### 2. 
$$
L = \left| \det \begin{pmatrix} 2 & 0 \\ 0 & 3 \end{pmatrix} \right| 
= |(2)(3) - (0)(0)| = |6| = \boxed{6}
$$

---

### 3. 
Vektor:
$
\vec{AB} = (1, 3, -1), \quad \vec{AC} = (2, 1, 1)
$

Cross product:
$$
\vec{AB} \times \vec{AC} = 
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
1 & 3 & -1 \\
2 & 1 & 1
\end{vmatrix} 
= (3 \cdot 1 - (-1) \cdot 1)\mathbf{i} 
- (1 \cdot 1 - (-1) \cdot 2)\mathbf{j} 
+ (1 \cdot 1 - 3 \cdot 2)\mathbf{k}
$$

$$
= (3 + 1)\mathbf{i} - (1 + 2)\mathbf{j} + (1 - 6)\mathbf{k} 
= 4\mathbf{i} - 3\mathbf{j} - 5\mathbf{k}
$$

Magnitude:
$$
|\vec{AB} \times \vec{AC}| = \sqrt{4^2 + (-3)^2 + (-5)^2} = \sqrt{16 + 9 + 25} = \sqrt{50}
$$

Luas segitiga:
$$
\text{Luas} = \frac{1}{2} \cdot \sqrt{50} = \boxed{\frac{\sqrt{50}}{2}}
$$

---

### 4. 
Vektor:
$
\vec{AB} = (-2, 4, 3), \quad \vec{AC} = (-4, -2, 5)
$

Cross product:
$$
\vec{AB} \times \vec{AC} = 
\begin{vmatrix}
\mathbf{i} & \mathbf{j} & \mathbf{k} \\
-2 & 4 & 3 \\
-4 & -2 & 5
\end{vmatrix} 
= (4 \cdot 5 - 3 \cdot (-2))\mathbf{i} 
- (-2 \cdot 5 - 3 \cdot (-4))\mathbf{j} 
+ (-2 \cdot (-2) - 4 \cdot (-4))\mathbf{k}
$$

$$
= (20 + 6)\mathbf{i} - (-10 + 12)\mathbf{j} + (4 + 16)\mathbf{k} 
= 26\mathbf{i} - 2\mathbf{j} + 20\mathbf{k}
$$

Magnitude:
$$
|\vec{AB} \times \vec{AC}| = \sqrt{26^2 + (-2)^2 + 20^2} = \sqrt{676 + 4 + 400} = \sqrt{1080}
$$

Luas segitiga:
$$
\text{Luas} = \frac{1}{2} \cdot \sqrt{1080} = \boxed{\frac{\sqrt{1080}}{2}}
$$
