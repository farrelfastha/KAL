```python
def creatdequeue():
    d=[]
    return d

def addfront(d,data):
    d.insert(0,data)
    return d

def addrear(d,data):
    d.append(data)
    return d

def removefront(d):
    data = d.pop(0)
    return data

def removerear(d):
    data = d.pop()
    return data

def size(d):
    return len(d)


def proses(data):
    antrian = creatdequeue()
    for nama,umur in data:
        if umur > 60:
            addfront(antrian,(nama,umur))

        else:
            addrear(antrian,(nama,umur))


    print(f"urutan layanan : ",end=" ")
    for nama,umur in antrian:
        print(f"{nama}",end=", ")



datapasien = [
    ("andi", 25),
    ("budi", 63),
    ("citra", 45),
    ("dedi", 70),
    ("ema",30)
]
proses(datapasien)
```
